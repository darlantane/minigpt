```python
import torch
from torch.utils.data import DataLoader

from tokenizer import CharTokenizer
from dataset import GPTDataset
from model import BigramLanguageModel

text = """
hello world
hello gpt
hello neural networks
"""
```

```python
tokenizer = CharTokenizer(text)

tokens = tokenizer.encode(text)

print(tokens[:20])
```

```python
block_size = 8

dataset = GPTDataset(
    tokens,
    block_size
)

len(dataset)
```

```python
x, y = dataset[0]

print(x)
print(y)

print(tokenizer.decode(x.tolist()))
print(tokenizer.decode(y.tolist()))
```

```python
loader = DataLoader(
    dataset,
    batch_size=4,
    shuffle=True
)
```

```python
xb, yb = next(iter(loader))

print(xb.shape)
print(yb.shape)
```

```python
model = BigramLanguageModel(
    tokenizer.vocab_size
)
```

```python
logits, loss = model(
    xb,
    yb
)

print(loss)
```

```python
optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=1e-2
)
```

```python
optimizer.zero_grad()

logits, loss = model(
    xb,
    yb
)

loss.backward()

optimizer.step()

print(loss.item())
```

```python
epochs = 100

for epoch in range(epochs):

    for xb, yb in loader:

        logits, loss = model(xb, yb)

        optimizer.zero_grad()

        loss.backward()

        optimizer.step()

    if epoch % 10 == 0:
        print(epoch, loss.item())
```

```python
start = torch.zeros(
    (1, 1),
    dtype=torch.long
)

generated = model.generate(
    start,
    100
)

print(
    tokenizer.decode(
        generated[0].tolist()
    )
)
```
