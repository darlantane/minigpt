```python
import torch

from model import BigramLanguageModel
```

```python
text = "hello world"

tokenizer = CharTokenizer(text)

tokens = tokenizer.encode(text)

dataset = GPTDataset(
    tokens,
    block_size=5
)
```

```python
model = BigramLanguageModel(
    tokenizer.vocab_size
)
```

```python
x, y = dataset[0]

x = x.unsqueeze(0)
y = y.unsqueeze(0)

print(x.shape)
```

```python
logits, loss = model(x, y)

print(logits.shape)
print(loss)
```

```python
start = torch.zeros(
    (1,1),
    dtype=torch.long
)

generated = model.generate(
    start,
    20
)

print(
    tokenizer.decode(
        generated[0].tolist()
    )
)
```
